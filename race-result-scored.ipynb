```python
%sql

CREATE OR REPLACE TABLE `drive-to-survive`.default.race_result_scored AS
SELECT 
  r.season,
  r.round,
  r.race_name,
  r.race_date,
  r.circuit_id,
  r.circuit_name,
  r.locality,
  r.country,
  r.position,
  r.points,
  r.grid,
  r.laps,
  r.status,
  r.driver_id,
  r.driver_first_name,
  r.driver_last_name,
  r.driver_nationality,
  r.constructor_id,
  r.constructor_name,
  r.constructor_nationality,
  r.fastest_lap_rank,
  r.fastest_lap_lap,
  r.fastest_lap_time,
  s_gp.gp,
  s_fl.fastest_lap,
  s_exp.gp AS gp_expected
FROM `drive-to-survive`.default.race_results r
INNER JOIN `drive-to-survive`.default.scoring s_gp
  ON r.position = s_gp.position
LEFT JOIN `drive-to-survive`.default.scoring s_fl
  ON r.fastest_lap_rank = s_fl.position
LEFT JOIN `drive-to-survive`.default.scoring s_exp
  ON r.grid = s_exp.position;
```
