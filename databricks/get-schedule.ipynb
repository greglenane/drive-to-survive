```python
import requests
import pandas as pd

url = f"https://api.jolpi.ca/ergast/f1/2025/races/"
response = requests.get(url)
data = response.json()

df = pd.json_normalize(
    data,
    record_path=["MRData", "RaceTable", "Races"],
    meta=[
        ["MRData", "RaceTable", "season"]
    ]
)

df.columns = (
    df.columns
      .str.replace(".", "_", regex=False)
      .str.lower()
)

df_spark = spark.createDataFrame(df)

df_spark.write.mode("overwrite").saveAsTable("`drive-to-survive`.default.schedule")
```
