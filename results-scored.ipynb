```python
%sql

CREATE OR REPLACE TABLE results_scored AS
SELECT
  rrs.season,
  rrs.round,
  rrs.race_name,
  rrs.race_date,
  rrs.circuit_id,
  rrs.circuit_name,
  rrs.locality,
  rrs.country,
  rrs.position,
  rrs.points,
  rrs.grid,
  rrs.laps,
  rrs.status,
  rrs.driver_id,
  rrs.driver_first_name,
  rrs.driver_last_name,
  rrs.driver_nationality,
  rrs.constructor_id,
  rrs.constructor_name,
  rrs.constructor_nationality,
  rrs.fastest_lap_rank,
  rrs.fastest_lap_lap,
  rrs.fastest_lap_time,
  rrs.gp,
  rrs.fastest_lap,
  srs.position AS sprint_position,
  srs.points AS sprint_points,
  srs.time AS sprint_time,
  srs.sprint
FROM
  `drive-to-survive`.default.race_result_scored rrs
LEFT JOIN
  `drive-to-survive`.default.sprint_result_scored srs
ON
  rrs.driver_id = srs.driver_id
  AND rrs.round = srs.round
  AND rrs.season = srs.season;
```
