```python
%sql

CREATE OR REPLACE TABLE `drive-to-survive`.default.results_scored AS
SELECT
  rrs.season,
  rrs.round,
  rrs.race_name,
  rrs.race_date,
  rrs.circuit_id,
  rrs.circuit_name,
  rrs.locality,
  rrs.country,
  rrs.points AS gp_points,
  rrs.laps,
  rrs.status,
  rrs.driver_id,
  rrs.driver_first_name,
  rrs.driver_last_name,
  rrs.driver_nationality,
  rrs.constructor_id,
  rrs.constructor_name,
  rrs.constructor_nationality,
  rrs.fastest_lap_lap,
  rrs.fastest_lap_time,
  srs.points AS sprint_points,
  srs.time AS sprint_time,

  rrs.grid AS gp_grid,
  rrs.gp_expected,
  rrs.position AS gp_position,
  rrs.gp,
  rrs.fastest_lap_rank,
  rrs.fastest_lap,
  srs.grid AS sprint_grid,
  srs.sprint_expected,
  srs.position AS sprint_position,
  srs.sprint,
  (COALESCE(rrs.gp, 0) + COALESCE(rrs.fastest_lap, 0) + COALESCE(srs.sprint, 0)) AS total
FROM
  `drive-to-survive`.default.race_result_scored rrs
LEFT JOIN
  `drive-to-survive`.default.sprint_result_scored srs
ON
  rrs.driver_id = srs.driver_id
  AND rrs.round = srs.round;
```
