```python
%sql

CREATE OR REPLACE TABLE `drive-to-survive`.default.sprint_scored AS
SELECT
  sp.Name,
  sp.Round,
  sp.Date,
  sp.Race,
  sp.Driver,
  sp.season,
  sp.sprint_round,
  sp.race_name,
  sp.points,
  sp.driver_id,
  sp.given_name,
  sp.family_name,
  sp.constructor_id,
  sp.time,
  s.*
FROM `drive-to-survive`.default.picks_sprint_results sp
INNER JOIN `drive-to-survive`.default.scoring s
  ON sp.position = s.position
WHERE s.position IS NOT NULL;
```
