```python
%sql

CREATE OR REPLACE TABLE `drive-to-survive`.default.picks_scored AS
SELECT
  rs.Name,
  rs.Round,
  rs.Date,
  rs.Race,
  rs.Driver,
  rs.season,
  rs.race_round,
  rs.race_name,
  rs.race_date,
  rs.circuit_id,
  rs.circuit_name,
  rs.locality,
  rs.country,
  rs.points AS rs_points,
  rs.grid,
  rs.laps,
  rs.status,
  rs.driver_id,
  rs.driver_first_name,
  rs.driver_last_name,
  rs.driver_nationality,
  rs.constructor_id,
  rs.constructor_name,
  rs.constructor_nationality,
  rs.fastest_lap_rank,
  rs.fastest_lap_lap,
  rs.fastest_lap_time,
  rs.position AS rs_position,
  rs.gp,
  rs.fastest_lap AS rs_fastest_lap,
  ss.points AS ss_points,
  ss.time AS ss_time,
  ss.position AS ss_position,
  ss.sprint AS ss_sprint
FROM `drive-to-survive`.default.results_scored rs
LEFT JOIN `drive-to-survive`.default.sprint_scored ss
  ON rs.Name = ss.Name AND rs.Round = ss.Round;
```
