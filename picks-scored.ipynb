```python
%sql 

CREATE OR REPLACE TABLE `drive-to-survive`.default.picks_scored AS
SELECT
  p.Name,
  p.Round,
  p.Date,
  p.Race,
  p.Driver,
  r.season,
  r.race_name,
  r.race_date,
  r.circuit_id,
  r.circuit_name,
  r.locality,
  r.country,
  r.gp_points,
  r.laps,
  r.status,
  r.driver_id,
  r.driver_first_name,
  r.driver_last_name,
  r.driver_nationality,
  r.constructor_id,
  r.constructor_name,
  r.constructor_nationality,
  r.fastest_lap_lap,
  r.fastest_lap_time,
  r.sprint_points,
  r.sprint_time,

  r.gp_grid,
  r.gp_expected,
  r.gp_position,
  r.gp,
  r.gp_var,
  r.fastest_lap_rank,
  r.fastest_lap,
  r.sprint_grid,
  r.sprint_expected,
  r.sprint_position,
  r.sprint,
  r.sprint_var,
  r.total,
  r.total_expected,
  r.total_var
FROM
  `drive-to-survive`.default.picks p
INNER JOIN
  `drive-to-survive`.default.results_scored r
ON
  p.Round = r.Round
  AND p.Driver = concat(r.driver_first_name, ' ', r.driver_last_name)
```
