```python
from pyspark.sql.functions import col, explode
df = spark.table("`drive-to-survive`.default.sprint_response")
```

```python
df_races = df.select(
    col("`MRData.RaceTable.season`").alias("season"),
    col("`MRData.RaceTable.round`").alias("round"),
    explode(col("`MRData.RaceTable.Races`")).alias("race")
)
```

```python

df_sprint = df_races.select(
        "season",
        "round",
        col("race.raceName").alias("race_name"),
        explode(col("race.SprintResults")).alias("sprint")
    )

```

```python
df_sprint_flat = df_sprint.select(
    "season",
    "round",
    "race_name",
    col("sprint.grid").cast("int").alias("grid"),
    col("sprint.position").cast("int").alias("position"),
    col("sprint.points").cast("int").alias("points"),
    col("sprint.Driver.driverId").alias("driver_id"),
    col("sprint.Driver.givenName").alias("given_name"),
    col("sprint.Driver.familyName").alias("family_name"),
    col("sprint.Constructor.constructorId").alias("constructor_id"),
    col("sprint.Time.time").alias("time")
)
```

```python
df_sprint_flat.write.mode("overwrite").saveAsTable("`drive-to-survive`.default.sprint_results")
```
