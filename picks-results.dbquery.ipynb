```sql
CREATE OR REPLACE TABLE `drive-to-survive`.default.picks_results AS
SELECT
  p.Name,
  p.Round,
  p.Date,
  p.Race,
  p.Driver,
  rr.season,
  rr.round AS race_round,
  rr.race_name,
  rr.race_date,
  rr.circuit_id,
  rr.circuit_name,
  rr.locality,
  rr.country,
  rr.position,
  rr.points,
  rr.grid,
  rr.laps,
  rr.status,
  rr.driver_id,
  rr.driver_first_name,
  rr.driver_last_name,
  rr.driver_nationality,
  rr.constructor_id,
  rr.constructor_name,
  rr.constructor_nationality,
  rr.fastest_lap_rank,
  rr.fastest_lap_lap,
  rr.fastest_lap_time
FROM `drive-to-survive`.default.picks p
LEFT JOIN `drive-to-survive`.default.race_results rr
  ON p.Round = rr.round
  AND p.Driver = concat(rr.driver_first_name, ' ', rr.driver_last_name);
```
