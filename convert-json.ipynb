```python
from pyspark.sql.functions import col, explode
df = spark.table("`drive-to-survive`.default.api_response")
```

```python
df_flat = df.select(
    col("`MRData.RaceTable.season`").alias("season"),
    col("`MRData.RaceTable.round`").alias("round"),
    explode(col("`MRData.RaceTable.Races`")).alias("Race")
)
```

```python

df_race = df_flat.select(
    "season",
    "round",

    col("Race.raceName").alias("race_name"),
    col("Race.date").alias("race_date"),

    col("Race.Circuit.circuitId").alias("circuit_id"),
    col("Race.Circuit.circuitName").alias("circuit_name"),

    col("Race.Circuit.Location.locality").alias("locality"),
    col("Race.Circuit.Location.country").alias("country"),

    explode("Race.Results").alias("Result")
)
```

```python
df_results = df_race.select(
    "season",
    "round",
    "race_name",
    "race_date",
    "circuit_id",
    "circuit_name",
    "locality",
    "country",

    col("Result.position").alias("position"),
    col("Result.points").alias("points"),
    col("Result.grid").alias("grid"),
    col("Result.laps").alias("laps"),
    col("Result.status").alias("status"),

    col("Result.Driver.driverId").alias("driver_id"),
    col("Result.Driver.givenName").alias("driver_first_name"),
    col("Result.Driver.familyName").alias("driver_last_name"),
    col("Result.Driver.nationality").alias("driver_nationality"),

    col("Result.Constructor.constructorId").alias("constructor_id"),
    col("Result.Constructor.name").alias("constructor_name"),
    col("Result.Constructor.nationality").alias("constructor_nationality"),

    col("Result.FastestLap.rank").alias("fastest_lap_rank"),
    col("Result.FastestLap.lap").alias("fastest_lap_lap"),
    col("Result.FastestLap.Time.time").alias("fastest_lap_time")
)
```

```python
df_results.write.mode("overwrite").saveAsTable("`drive-to-survive`.default.race_results")
```
