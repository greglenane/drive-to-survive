```sql
CREATE OR REPLACE TABLE `drive-to-survive`.default.picks_sprint_results AS
SELECT
  p.Name,
  p.Round,
  p.Date,
  p.Race,
  p.Driver,
  sr.season,
  sr.round AS sprint_round,
  sr.race_name,
  sr.position,
  sr.points,
  sr.driver_id,
  sr.given_name,
  sr.family_name,
  sr.constructor_id,
  sr.time
FROM `drive-to-survive`.default.picks p
LEFT JOIN `drive-to-survive`.default.sprint_results sr
  ON p.Round = sr.round
  AND p.Driver = concat(sr.given_name, ' ', sr.family_name);
```
