```sql
CREATE OR REPLACE TABLE `drive-to-survive`.default.results_scored AS
SELECT 
  p.Name,
  p.Round,
  p.Date,
  p.Race,
  p.Driver,
  p.season,
  p.race_round,
  p.race_name,
  p.race_date,
  p.circuit_id,
  p.circuit_name,
  p.locality,
  p.country,
  p.points,
  p.grid,
  p.laps,
  p.status,
  p.driver_id,
  p.driver_first_name,
  p.driver_last_name,
  p.driver_nationality,
  p.constructor_id,
  p.constructor_name,
  p.constructor_nationality,
  p.fastest_lap_rank,
  p.fastest_lap_lap,
  p.fastest_lap_time,
  s.position,
  s.gp,
  s.sprint,
  s.fastest_lap
FROM `drive-to-survive`.default.picks_results p
INNER JOIN `drive-to-survive`.default.scoring s
  ON p.position = s.position
WHERE s.position IS NOT NULL
```
